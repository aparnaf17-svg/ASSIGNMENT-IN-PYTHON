```python
import numpy as np

# Step 1: Create 1D NumPy array for Week 1
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

# Step 2: Inspection and Properties
print("--- Week 1 Temperature Inspection ---")
print("Array:", temperatures_w1)
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements (Size):", temperatures_w1.size)
```

```python
# Convert Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32
print("Temperatures in Fahrenheit:", temperatures_f)

# Calculate statistics
max_temp = np.max(temperatures_w1)
min_temp = np.min(temperatures_w1)
mean_temp = np.mean(temperatures_w1)

print(f"Maximum Temperature: {max_temp}°C")
print(f"Minimum Temperature: {min_temp}°C")
print(f"Mean Temperature: {mean_temp:.2f}°C")
```

```python
# First three days (indices 0 to 2)
first_three = temperatures_w1[:3]

# Weekend / Last two days (indices 5 and 6)
weekend = temperatures_w1[-2:]

# Middle three days (indices 2 to 4: days 3, 4, and 5)
middle_three = temperatures_w1[2:5]

print("First three days:", first_three)
print("Weekend temperatures:", weekend)
print("Middle three days:", middle_three)
```

```python
# Step 5: Create 2D NumPy array
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]   # Week 2
])

# Step 6: Inspect 2D Array
print("--- 2D Temperatures Inspection ---")
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)

# Slice each week (rows)
week1 = temperatures[0, :]
week2 = temperatures[1, :]

# Slice weekends (last two columns for both weeks)
weekends_both = temperatures[:, -2:]

print("Week 1 Temperatures:", week1)
print("Week 2 Temperatures:", week2)
print("Weekends for both weeks:\n", weekends_both)
```

```python
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)
print("--- Original Marks Series ---")
print(marks)
```

```python
# Integer index position for 1st rank student (position 0)
print("1st Rank Mark (via position 0):", marks.iloc[0])

# loc accessor to retrieve top 3 ranks
print("\nTop 3 Ranks:\n", marks.loc[['Rank1', 'Rank2', 'Rank3']])

# iloc accessor to retrieve the mark of the 3rd rank student (position 2)
print("\n3rd Rank Mark (via position 2):", marks.iloc[2])

# Boolean mask for marks greater than 90
high_scorers = marks[marks > 90]
print("\nRanks with marks > 90:\n", high_scorers)
```

```python
# Modify 1st rank to 100
marks['Rank1'] = 100

# Remove the entry corresponding to the last rank ('Rank5')
marks = marks.drop('Rank5')

# Compute CGPA by dividing each mark by 10
cgpa = marks / 10

print("--- Modified Marks ---")
print(marks)
print("\n--- Computed CGPA ---")
print(cgpa)
```

```python
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 
                        'Clothing', 'Electronics', 'Furniture', 'Clothing', 
                        'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 
               'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)
```

```python
print("--- Head ---")
print(transactions.head())

print("\n--- Tail ---")
print(transactions.tail())

print("\n--- Shape, Columns, and Info ---")
print("Shape:", transactions.shape)
print("Columns:", transactions.columns.tolist())
print(transactions.info())

# Display only ProductCategory and Amount
print("\n--- ProductCategory & Amount ---")
print(transactions[['ProductCategory', 'Amount']])

# Retrieve the last 3 columns using iloc
print("\n--- Last 3 Columns ---")
print(transactions.iloc[:, -3:])

# Filter rows where Region is 'North' and Amount > 200
filtered_trans = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
print("\n--- Filtered Transactions (North & Amount > 200) ---")
print(filtered_trans)

# Value counts for ProductCategory
print("\n--- Product Category Counts ---")
print(transactions['ProductCategory'].value_counts())

# Unique values in Region
print("\n--- Unique Regions ---")
print(transactions['Region'].unique())

# Group by Region and find mean amount
print("\n--- Mean Amount by Region ---")
print(transactions.groupby('Region')['Amount'].mean())
```

```python
# Modify Amount for TransactionID 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

# Add a new column 'Discount' (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10

# Remove the row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]

# Delete the 'Discount' column
transactions = transactions.drop(columns=['Discount'])

print("--- Final Manipulated DataFrame ---")
print(transactions)
```
